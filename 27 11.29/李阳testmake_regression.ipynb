```python
import numpy as np
#my draft
def my_make_regression(n_samples,n_features,bias=0,coef=False,noise=0,random_state=None):
    np.random.seed(random_state)
    X=np.random.randn(n_samples,n_features)
    w=np.random.randn(n_features)*100
    y=X@w+bias
    if noise>0:
        y= y+np.random.randn(n_samples)*noise
    if coef is True:
        return X,y,w
    else:
        return X,y
```

```python
x,y,w=my_make_regression(n_samples=10,n_features=3,bias=20,noise=5,coef=True,random_state=1)
x,y,w
```

```python
#ai optimized definitive version
def my_make_regression(n_samples, n_features, noise=0.0, coef=False):
    import numpy as np
    
    # 参数检查，用 assert
    assert isinstance(n_samples, int) and n_samples > 0, "n_samples 必须为正整数"
    assert isinstance(n_features, int) and n_features > 0, "n_features 必须为正整数"
    assert isinstance(noise, (int, float)) and noise >= 0, "noise 必须为非负数"
    assert isinstance(coef, bool), "coef 必须是布尔值 True 或 False"
    
    X = np.random.randn(n_samples, n_features)
    w = np.random.randn(n_features)
    y = X @ w
    
    if noise > 0:
        y += np.random.randn(n_samples) * noise
    
    return (X, y, w) if coef else (X, y)
```
