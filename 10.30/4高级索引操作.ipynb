```python
import pandas as pd
import numpy as np

np.random.seed(0)

df=pd.DataFrame(    (np.random.randint(0,20,(4,4))),
                index=[['上半年','上半年','下半年','下半年'],['A组','B组','A组','B组']],
                columns=[['一区','一区','二区','二区'],['苹果','橘子','苹果','橘子']]
                )
df
```

```python
#列操作
df['一区']
df['一区','橘子']
```

```python
#行操作
df.loc['上半年']
df.loc['上半年','A组']
df.loc[['上半年','下半年']]
```

```python
df.sort_index()#默认最外层
df.sort_index(level=1)#按内层索引排序
df.sort_index(axis=1)
```

```python
#索引d堆叠
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, np.nan],
    'B': [3, 4]
}, index=['x', 'y'])

print(df)
df.stack()
df.stack(dropna=False)


```

```python
df.unstack()
```

```python
import pandas as pd

# 多层索引
index = pd.MultiIndex.from_tuples([('A', 1), ('A', 2), ('B', 1)], names=['group', 'id'])
data = pd.DataFrame({'value':[10, 20, 30]}, index=index)
print(data)
df_unstacked = data.unstack(level=-1)
print(df_unstacked)
df_unstacked = data.unstack(level=-1, fill_value=0)
print(df_unstacked)


```
