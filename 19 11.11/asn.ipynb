```python
import pandas as pd

# 示例数据
data = [
    [1, 2, '2016-03-01', 5],
    [1, 2, '2016-05-02', 6],
    [1, 3, '2017-06-25', 1],
    [3, 1, '2016-03-02', 0],
    [3, 4, '2018-07-03', 5]
]

Activity = pd.DataFrame(data, columns=['player_id', 'device_id', 'event_date', 'games_played'])

# 转换日期列类型（以确保排序正确）
Activity['event_date'] = pd.to_datetime(Activity['event_date'])

# 按玩家和日期排序
Activity = Activity.sort_values(['player_id', 'event_date'])

# 按 player_id 分组，对 games_played 求累计和
Activity['games_played_so_far'] = Activity.groupby('player_id')['games_played'].cumsum()

# 选取指定列输出
result = Activity[['player_id', 'event_date', 'games_played_so_far']]

print(result)
```

```python
import pandas as pd

# 示例数据
data = [
    [1, 2, '2016-03-01', 5],
    [1, 2, '2016-03-02', 6],
    [2, 3, '2017-06-25', 1],
    [3, 1, '2016-03-02', 0],
    [3, 4, '2018-07-03', 5]
]

Activity = pd.DataFrame(data, columns=['player_id', 'device_id', 'event_date', 'games_played'])
Activity['event_date'] = pd.to_datetime(Activity['event_date'])

# 1️⃣ 每个玩家的首次登录日期
first_login = Activity.groupby('player_id', as_index=False)['event_date'].min().rename(columns={'event_date':'first_login'})

# 2️⃣ 合并回原表
df = Activity.merge(first_login, on='player_id')

# 3️⃣ 筛选出“首次登录后第二天登录”的玩家
df_next_day = df[df['event_date'] == df['first_login'] + pd.Timedelta(days=1)]

# 4️⃣ 计算比例
total_players = Activity['player_id'].nunique()
players_next_day = df_next_day['player_id'].nunique()
fraction = round(players_next_day / total_players, 2)

# 5️⃣ 输出结果格式
result = pd.DataFrame({'fraction': [fraction]})
print(result)
```

```python
import pandas as pd

def find_winner(candidate: pd.DataFrame, vote: pd.DataFrame) -> pd.DataFrame:
    # 统计每个候选人的票数
    vote_count = vote.groupby('candidateId').size().reset_index(name='votes')

    # 找出票数最多的候选人 id
    winner_id = vote_count.loc[vote_count['votes'].idxmax(), 'candidateId']

    # 在 candidate 表中查找该候选人的名字
    result = candidate[candidate['id'] == winner_id][['name']]

    return result
```
