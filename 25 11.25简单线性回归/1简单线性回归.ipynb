```python
#画常见函数图像
import numpy as np
import matplotlib.pyplot as plt
#生成从start到stop，num个元素的等差数列
x=np.linspace(0,3,100) 
print(x)
#创建画布与子图
fig,ax=plt.subplots(2,2)
ax[0,0].plot(x,np.exp(x)+8*x)
ax[0,1].plot(x,x**2-6*x+3)
ax[1,0].plot(x,8*x-2)
ax[1,1].plot(x,np.sin(x))
#若是觉得二维数组麻烦，可以使用ravel降维成一维数组，然后[0],[1],[2]……来索引
# ax=ax.ravel()
# ax[0].plot(x,np.exp(x)+8*x)
# ax[1].plot(x,x**2-6*x+3)
# ax[2].plot(x,8*x-2)
# ax[3].plot(x,np.sin(x))
```

```python
#使用make_regression生成随机的数据集
from sklearn.datasets import make_regression
x,y,w=make_regression(n_samples=100,n_features=1,coef=True,
                      bias=0,noise=2.5,random_state=23)
print(w)
print(x)
print(y)#X（特征）永远是二维矩阵 → shape = (样本数, 特征数),y（标签）默认是一维向量 → shape = (样本数,)
#手动计算y
print(x.ravel()*w+2)
#画xy的散点图
plt.scatter(x,y, marker=".")
```

```python
#尝试自己洗牌
a = np. arange (10)#生成array range，起始值默认为0，终止值自设，步长默认为1
b = a * 20
np. random. seed (30)
np. random. shuffle (a)
np. random. seed (30)#若去掉该步，则ab各自洗牌，结果不一一对应
np. random. shuffle (b)
print (a)
print (b)
```

```python
#通过索引来实现洗牌后顺序依旧对应
a = np. arange (10)
b = a * 20
index=np.arange(a.shape[0])
np.random.shuffle(index)
print(index)
print(a[index])
print(b[index])
```

```python
#使用train_test_split切分数据集
#切分时，需多个数组的第一维长度（样本数量）相同
from sklearn.model_selection import train_test_split
#test和train_size直写一个就行：整数参数时，表示绝对的样本数量；浮点数时，表示比例
x_train,x_test,y_train,y_test=train_test_split(x.ravel(),y,
                                               test_size=0.3,
shuffle=True,random_state=3)#前面用make_regression生成的x是2维的，需要ravel一下
#即使种子固定，shuffle也会运行一次，也就是说，数据会和传入时顺序不同
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
# print(x_train)
# print(x_test)
# print(y_train)
# print(y_test)
```

```python
#参数求解
a=np.sum((x_train-x_train.mean())*(y_train-y_train.mean()))#该步运算逻辑如下
#步骤 1：计算x_train的均值（x_train.mean()）
#步骤 2：计算x_train的离均差（x_train - x_train.mean()）
#步骤 3：计算y_train的均值和离均差（y_train - y_train.mean()）
#步骤 4：对应元素相乘（(x_train - x_train.mean()) * (y_train - y_train.mean())）
#步骤 5：求和（np.sum(...)）
# 可以写成a=(x_train-x_train.mean())@(y_train-y_train.mean())，即直接点积
b=np.sum((x_train-x_train.mean())**2)
w1=a/b
w0=y_train.mean()-w1*x_train.mean()
print(w1)
print(w0)
```

```python
#参数用点积计算
c=x_train-x_train.mean()
d=y_train-y_train.mean()
a=np.dot(c,d)
b=np.dot(c,c)
w1=a/b
w0=y_train.mean()-w1*x_train.mean()
print(w1)
print(w0)
```

```python
#实现预测
y_hat=w1*x_test+w0
print(y_hat)
print(y_test)
```

```python
#绘制回归线
# plt.rcParams["font.family"] = "SimHei"
# plt.rcParams["axes.unicode_minus"] = False
plt.scatter(x_train, y_train, c="g", marker=".", label="训练集")
plt.scatter(x_test,y_test, c="orange", marker="*", label="测试集")
t = np.linspace(x_train.min() - 0.5, x_train.max() + 0.5, 100)#密集划分点
t2 = w0 + w1 * t
plt.plot(t, t2, c="r", label="回归线")#将点连线
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"回归方程：y={w0:.2f} + {w1:.2f} * x")
```

```python
#绘制测试集上真实值与预测值的对比
plt.figure(figsize=(12, 5))
index = np.arange(y_test.shape[0])
plt.plot(index,y_test, label="真实值", color="r", marker="o", ls="--")
plt.plot(index, y_hat, label="预测值", color="g", marker="o", ls="--")
plt.xlabel("测试集样本序号")
plt.ylabel("样本的真实值（预测值）")
plt.legend()
```

```python
# 使用sklearn来学习模型
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
#sklearn中，fit要求X为2维数组
x_train=x_train.reshape(-1,1)
x_test=x_test.reshape(-1,1)
#使用训练集训练模型
lr.fit(x_train,y_train)
print(lr.coef_)
print(lr.intercept_)
#使用模型对未知样本预测
y_hat=lr.predict(x_test)
print(y_hat)
```
