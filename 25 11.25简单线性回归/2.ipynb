```python
train_x,test_x,train_y,test_y=train_test_split(x.ravel(),y,test_size=0.3,shuffle=True,random_state=0)
print(train_x.shape,test_x.shape,train_y.shape,test_y.shape)
```

```python
a=np.sum((train_x-train_x.mean())*(train_y-train_y.mean()))
b=np.sum((train_x-train_x.mean())**2)
w1=a/b
w0=train_y.mean()-w1*train_x.mean()
a
b
w1
w0
```

```python
y_hat=w1*test_x+w0
y_hat
test_y
```

```python
plt.scatter(train_x,train_y,c='g',marker='.',label="训练集")
plt.scatter(test_x,test_y,c='orange',marker='*',label="测试集")
t=np.linspace(train_x.min()-0.5,train_x.max()+0.5,100)
t_hat=w1*t+w0
plt.plot(t,t_hat,c='r',label='回归线')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.title(f'回归方程：y={w0:.2f}+{w1:.2f}*x')
```

```python
plt.figure(figsize=(12,5))
index=np.arange(test_y.shape[0])
plt.plot(index,test_y,label='tv',c='r',marker='o')
plt.plot(index,y_hat,label='pv',c='g',marker='o',ls='--')
plt.legend()
plt.xlabel('index')
plt.ylabel('tv and pv')
```
