```python
import numpy as np
import pandas as pd
import seaborn as sns
from statsmodels.stats.weightstats import ztest
from scipy.stats import chi2_contingency
df_train=pd.read_csv('../data/new_car_train.csv')
df_train
```

```python
arr=[]
for col in df_train:
    if df_train[col].nunique()>30:
        p_value=ztest(df_train.query('价格标签==0')[col],df_train.query('价格标签==1')[col])[1]
        # p_value
        arr.append(
            pd.DataFrame({
                "维度":[col],
                '检验方法':['z检验'],
                'p值':[p_value],
                '结果':['有影响' if p_value<0.05 else '没影响']
            })
        )
    else:
        p_value=chi2_contingency(df_train.groupby([col,'价格标签']).size().unstack())[1]
        arr.append(
            pd.DataFrame({
                "维度":[col],
                '检验方法':['卡方检验'],
                'p值':[p_value],
                '结果':['有影响' if p_value<0.05 else '没影响']
            })
        )
pd.concat(arr)
```
