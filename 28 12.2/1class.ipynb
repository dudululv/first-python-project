```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
class P:
    def __init__(self,name='gump',age=15) -> None:#定义初始化方法,类的对象,具有哪些属性,可以在该方法中定义并初始化,self参数是保留参数,表示当前创建的对象
    #可以通过在init中传参数的方法来设置初始化  
        print('creat object')
        self.name=name
        self.age=age
    def walk (self):
        return f'{self.name} is run'
    
p=P()
p.name
p2=P('susan',21)
p2.name
p.__init__()
p.walk()
```

```python
class A:
    def __init__(self, name):
        self.name = name

    def greet(self, other):
        print(self.name, "向", other.name, "打招呼")
a = A("小明")
b = A("小红")

a.greet(b)
```

```python
class B:
    def __init__(self, value):
        self.value = value


class A:
    def __init__(self, other_obj):
        self.other = other_obj  # 它等价于：在对象的“内部字典”里添加一个键值对："other": other_obj

    def show(self):
        print(self.other.value)  # 直接用另一个对象
b = B(100)
a = A(b)

a.show()
```

```python
class C:
    def __init__(self):
        self.x = 42


class D:
    def __init__(self):
        self.c = C()  # 直接创建另一个类的对象

    def use_c(self):
        print(self.c.x)
d = D()
d.use_c()
```

```python
#init在创建对象时执行
#str将对象转化为str类型时执行
#call将对象当成函数调用时执行
from typing import Any


class P:
    def __init__(self,name=None) -> None:
        print('creat ob')
        self.name=name
    def __str__(self) -> str:
        return f'i am {self.name}'
    def __call__(self, *args: Any, **kwds: Any) -> Any:
        print('def')
        return 'callback'
    def __len__(self):
        return 10
p=P()
r=p()
print(r)
len(r)
```

```python
#面向对象三大特征
#封装/继承/多态
class p:
    def __init__(self,name,age,k) -> None:
        self.name=name
        self.__age=age
        # self.__run()
        self._k=k#受保护对象,和正常对象完全一致
    def getage(self):#私有属性,在类的内部可以用
        return self.__age
    def __run(self):#私有方法,在类的内部可以用
        print('private')
r=p('g',11,2)
r.name
r.getage()
r._p__age#真实名称为_class__member
r._k
```
