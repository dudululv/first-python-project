```python
"""
继承发生在父类与子类之间。子类是一种特殊的父类。

子类继承父类，子类就会成为一种特殊的父类，子类就具有父类的一切属性与行为（能力）。
父类能做的子类也可以做。

在程序中，子类继承父类，子类就可以继承父类中定义的属性与方法，就好像在子类中自己定义的一样。

继承语法：

class 子类(父类1, 父类2, ……, 父类n):
    类体实现。
    
如果没有显示指定父类，则当前类继承object类。

问题：我们为什么要使用继承？
我们需要用到父类的时候，我们创建父类的对象就可以了。我们为什么要定义子类，使用子类的对象呢？

子类继承父类，子类就可以获取父类的能力（继承父类的属性与方法）。这就能够避免代码的重复，实现代码重用。
而且，当以后父类更新时，子类可以随父类的更新，自动完成更新。

但是，如果子类只是单纯的继承父类，而没有任何调整与改动，继承还是没有意义的。因为这种情况下，我们只需要
使用父类的对象就可以完成任务，根本无需使用子类对象。子类继承父类，除了吸收父类的功能以外，子类也需要具有
一定的创新能力。这体现为：增加新的属性或方法，或者对现有的父类功能进行调整与改进。

"""

class Person:
    def __init__(self, name):
        self.name = name
        
    def run(self):
        print("跑步")
        

class Student(Person):
    # 子类增加新的功能。
    def read_book(self):
        print("读书。")


s = Student("张三")
print(s.name)
s.run()
s.read_book()
```

```python
"""
子类继承父类，如果父类中存在不适合于子类的方法，子类可以重新定义父类的方法，这称为方法重写。

如果子类重写了父类的方法，则通过子类对象调用的，将是子类自己定义的方法，而不是从父类中继承的方法。
"""

class Person:
    def run(self):
        print("跑步")
        
class Baby(Person):
    # 重写父类中的方法。
    def run(self):
        print("暂时还不能跑步。")
        
        
b = Baby()
b.run()
```

```python
"""
子类继承父类：
1. 如果父类的方法适合于子类，则子类直接继承父类的方法。
2. 如果父类的方法完全不适合子类，则子类可以重写父类的方法。
3. 如果父类的方法并没有完全适于子类，但又没有完全不适合于子类（部分适合于子类），则子类重写父类的方法，
   对于父类中适合的部分，在子类中调用父类的方法实现，对于不适合子类的部分，子类单独给出实现。
"""

class Person:
    def __init__(name, age):
        self.name = name
        self.age = age
        
        
class Student(Person):
    def __init__(name, age, code):
#         self.name = name
#         self.age = age
        # name与age属性的初始化，是父类中具有的功能，这部分功能适合于子类，因此，
        # 我们在子类中调用父类的方法来实现。
        super().__init__(name, age)
        # code的初始化，父类中不没有提供，则子类单独给出实现。
        self.code = code
```

```python
# Python 的多态不靠语法约束，而是靠“行为相同”。看起来不明显，但无处不在。
class Person:
    def walk(self):
        print("人在行走")
        
class Student(Person):
    def walk(self):
        print("齐步走。")
        
class Soldier(Person):
    def walk(self):
        print("正步走")
        
def walk(p):
    p.walk()
walk(Person())
walk(Student())
walk(Soldier())
```

```python
"""
类中定义的属性：
1. 类属性
2. 实例属性

类属性绑定到当前类，由所有对象所共享。
实例属性绑定到当前对象，每个对象都会具有自己的实例属性，如果其中一个对象的实例属性发生改变，则
对其他对象没有影响。

属性访问
类属性通过当前类来访问。 类名.类属性名。   类属性也可以通过对象名来访问。=> 对象名.类属性  但是强烈不推荐这么做。
实例属性通过当前对象来访问。    对象名.实例属性名。

定义规则：
如果当前的属性需要由所有对象所共享，则应当定义为类属性，如果当前属性需要由每个对象所独享，则应当定义为实例属性。
"""

class Person:
    # 类属性定义。
    count = 0
    
    def __init__(self, name):
        # 实例属性。
        self.name = name
        Person.count += 1
        
        
p = Person("张三")
# p2 = Person("李四")
p.name = "张改名"
print(p.name)
# print(p2.name)
# 应该通过类名来访问类属性。
print(Person.count)
# 也可以通过对象名访问类属性，但是不要这么做。
print(p.count)
# 注意：这里并没有修改类属性，而是为当前对象增加一个实例属性，实例属性名也是count。
p.count = 10
# print(p.count)
# print(Person.count)
# 如果需要修改类属性，则需要通过类名访问。
# Person.count = 10
# print(Person.count)
print(Person.count)
```

```python
"""
类中可以定义三种方法：
1. 实例方法
2. 类方法
3. 静态方法

实例方法
实例方法的第一个参数为当前对象，习惯上使用self命名。
实例方法通过当前对象访问。 =》 对象名.实例方法名(参数)。

类方法
类方法的第一个参数为当前类，习惯上使用cls命名。类方法通过@classmethod修饰。
类方法通过当前类来访问。 =》 类名.类方法名(参数)。

静态方法
静态方法没有保留参数。静态方法通过@staticmethod修饰。
静态方法通过当前类来访问。 =》 类名.静态方法名(参数)。


定义场景：
1. 如果方法中，需要使用当前对象（访问当前对象的实例属性，或者调用其他的实例方法），则应当定义为实例方法。
2. 如果方法中，不需要访问当前对象，但是需要访问当前类（访问当前类的类属性，或者调用当前类其他的类方法），则应当定义为类方法。
3. 如果方法中，既不需要使用当前对象，也不需要使用当前类，则应当定义为静态方法。从这个角度看，静态方法与当前类就没有太多的关联，
   我们完全可以将静态方法定义在类的外部，成为一个函数。之所以定义在类的内部，称为静态方法，是因为逻辑上的统一（整体）。
"""

class Person:
    count = 0
    
    def __init__(self, name):
        self.name = name
        Person.count += 1
        
    def introduce(self):
        print(f"大家好，我是{self.name}。")
        
    @classmethod
    def get_count(cls):
        return cls.count
    
    @staticmethod
    def make_friend(p, p2):
        print(f"{p.name}与{p2.name}交朋友。")
        

p = Person("张三")
p2 = Person("李四")
# 实例方法，通过对象来访问。
p.introduce()
# 类方法与静态方法，通过类名来访问。
Person.make_friend(p, p2)
print(Person.get_count())
```
