```python
from sklearn.linear_model import SGDRegressor
from sklearn.datasets import make_regression
X,y,coef=make_regression(n_samples=1000,n_features=5,noise=2,bias=0,random_state=2323,coef=True)
print('real coef',coef)
#eta0:初始学习率
sgd=SGDRegressor(eta0=0.2)
sgd.fit(X,y)
print('拟合coef',sgd.coef_)
print('拟合bias',sgd.intercept_) 
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
data=pd.read_csv('../data/house.csv')
display(data.head())
X,y=data.iloc[:,:-1],data['MEDV']
lr=LinearRegression()
lr.fit(X,y)
print(lr.coef_)
print(lr.intercept_)
print(lr.score(X,y))
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
#将数据的各个特征转换成均值为0,标准差为1的形式
#内部计算:每个特征,(特征-特征均值)/特征标准差
a=np.array([5,3,8,4,6,-2])
print(a.mean())
b=a-a.mean()
print(b.mean())
#数组中每个元素乘除一个值,则该数组的标准差也会乘除该值
print(a.std())
c=a*2
print(c.std())
d=a/a.std()
print(d.std())
```

```python
s = StandardScaler()
# fit学习转换规则，该规则为每个特征列的均值与标准差。
# 学习的规则通过属性来存储。
s.fit(X)
# 从训练集中学习的均值与方差。
print(s.mean_)
print(s.var_)
# 利用训练集学习到的转换规则，就可以对任意的未知数据进行转换。
X_scale = s.transform(X)
df = pd.DataFrame(X_scale, columns=X.columns)
df.head()
```

```python
# 是否标准化，不会影响到线性回归的拟合优度（R^2值），但是会影响线性回归拟合的参数（权重与偏置）。
lr.fit(X_scale, y)
print(lr.coef_)
print(lr.intercept_)
print(lr.score(X_scale, y))
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
lr = LinearRegression()
lr.fit(X_train, y_train)
print("线性回归拟合结果：")
print(lr.score(X_train, y_train))
print(lr.score(X_test, y_test))

sgd = SGDRegressor(eta0=0.01)
sgd.fit(X_train, y_train)
print("梯度法拟合结果：")
print(sgd.score(X_train, y_train))
print(sgd.score(X_test, y_test))
```

```python
from sklearn.pipeline import Pipeline
pipe = Pipeline([("scale", StandardScaler()), 
                 ("sgd", SGDRegressor(eta0=0.01))])
pipe.fit(X_train, y_train)
print(pipe.score(X_train, y_train))
print(pipe.score(X_test, y_test))
```
