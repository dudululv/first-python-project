```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from seaborn import histplot,heatmap,violinplot,boxplot
df=pd.read_csv('../data/处理后的数据2.csv')
# 不同性别用户在二级分类上的购买量总和对比
a=df.query('性别<2')
a.性别=a.性别.apply(lambda x:'男'if x else'女')
b=a.groupby(['二级分类','性别']).购买数量.sum().unstack()
b.sum(axis=1)
b=b.assign(合计=b.sum(axis=1)).sort_values('合计',ascending=False).head(20).drop(columns='合计')
b.plot.bar(title='购买量总和对比柱状图')
```

```python
# 二级分类购买量与累计比例的统计
df=pd.read_csv('../data/处理后的数据2.csv')
df.购买日期=pd.to_datetime(df.购买日期)
a=df.groupby('二级分类').购买数量.sum().to_frame().sort_values(by='购买数量',ascending=False)
a=a.assign(累计销量=a.a.cumsum()).eval('累计比例=累计销量/购买数量.sum()*100')
a = a.query('累计比例<=81')
a
a.drop('累计销量',axis=1).plot.bar(subplots=True,figsize=(15,6))
```
