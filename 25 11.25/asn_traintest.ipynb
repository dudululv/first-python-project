```python
import numpy as np
import pandas as pd

def train_test_split_fast(X, y, test_size=0.2, random_state=None):
    """
    纯 NumPy 实现 train_test_split，支持输入 X/y 为 DataFrame/Series 或 ndarray
    """
    # 保存输入类型
    X_is_df = isinstance(X, pd.DataFrame)
    X_is_series = isinstance(X, pd.Series)
    y_is_series = isinstance(y, pd.Series)
    y_is_df = isinstance(y, pd.DataFrame)
    
    # 转成 NumPy
    X_np = X.values if X_is_df or X_is_series else X
    y_np = y.values if y_is_df or y_is_series else y

    n_samples = X_np.shape[0]

    if random_state is not None:
        np.random.seed(random_state)
    
    # 打乱索引
    indices = np.random.permutation(n_samples)
    
    test_count = int(n_samples * test_size)
    test_indices = indices[:test_count]
    train_indices = indices[test_count:]
    
    X_train_np = X_np[train_indices]
    X_test_np = X_np[test_indices]
    y_train_np = y_np[train_indices]
    y_test_np = y_np[test_indices]
    
    # 返回原始类型
    if X_is_df:
        X_train_np = pd.DataFrame(X_train_np, columns=X.columns)
        X_test_np = pd.DataFrame(X_test_np, columns=X.columns)
    elif X_is_series:
        X_train_np = pd.Series(X_train_np.flatten(), name=X.name)
        X_test_np = pd.Series(X_test_np.flatten(), name=X.name)
    
    if y_is_df:
        y_train_np = pd.DataFrame(y_train_np, columns=y.columns)
        y_test_np = pd.DataFrame(y_test_np, columns=y.columns)
    elif y_is_series:
        y_train_np = pd.Series(y_train_np.flatten(), name=y.name)
        y_test_np = pd.Series(y_test_np.flatten(), name=y.name)
    
    return X_train_np, X_test_np, y_train_np, y_test_np

# 测试
X = pd.DataFrame({'feature1': np.arange(10), 'feature2': np.arange(10,20)})
y = pd.Series(np.arange(10)*2)

X_train, X_test, y_train, y_test = train_test_split_fast(X, y, test_size=0.3, random_state=42)

print("X_train:\n", X_train)
print("X_test:\n", X_test)
print("y_train:\n", y_train)
print("y_test:\n", y_test)
```

```python
import numpy as np


def train_test_split(x,y,test_size,shuffle,random_state=None):
    n_samples = len(x)
    
    
    np.random.seed(random_state)

    indices = np.arange(n_samples)

    if shuffle:
        np.random.shuffle(indices)

    test_count = int(n_samples * test_size)
    
    test_indices = indices[:test_count]
    train_indices = indices[test_count:]

    x_train = x[train_indices]
    x_test = x[test_indices]
    y_train = y[train_indices]
    y_test = y[test_indices]
    
    return x_train, x_test, y_train, y_test
# 测试
x = np.arange(10)
y = x * 2

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, shuffle=True)

print("x_train:", x_train)
print("x_test:", x_test)
print("y_train:", y_train)
print("y_test:", y_test)
```
