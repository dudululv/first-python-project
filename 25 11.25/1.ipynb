```python
train_x,test_x,train_y,test_y=train_test_split(x.ravel(),y,test_size=0.3,shuffle=True,random_state=0)
print(train_x.shape,test_x.shape,train_y.shape)
print(test_x)
#X（特征）永远是二维矩阵 → shape = (样本数, 特征数),y（标签）默认是一维向量 → shape = (样本数,)
```

```python
a=np.sum((train_x-train_x.mean())*(train_y-train_y.mean()))
b=np.sum((train_x-train_x.mean()**2))
w1=a/b
w0=train_y.mean()-w1*train_x.mean()
print(w1)
print(w0)
y_hat=w1*test_x+w0
print(y_hat)
print(test_y)
```

```python
x, y, w = make_regression(n_samples=100, n_features=1, noise=1.5, bias=0, coef=True, random_state=0)
x_train, x_test, y_train, y_test = train_test_split(x.ravel(), y, test_size=0.3, random_state=0)

c = x_train - x_train.mean()
d = y_train - y_train.mean()
a = np.dot(c, d)
b = np.dot(c, c)
w1 = a / b
print(w1)
print(test_y)
```

```python
plt.rcParams["font.family"] = "SimHei"
plt.rcParams["axes.unicode_minus"] = False

plt.scatter(train_x, train_y, c="g", marker=".", label="训练集")
plt.scatter(test_x, test_y, c="orange", marker="*", label="测试集")
t = np.linspace(train_x.min() - 0.5, train_x.max() + 0.5, 100)
t2 = w0 + w1 * t
plt.plot(t, t2, c="r", label="回归线")
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"回归方程：y={w0:.2f} + {w1:.2f} * x")

plt.figure(figsize=(12, 5))
index = np.arange(test_y.shape[0])
plt.plot(index,test_y, label="真实值", color="r", marker="o", ls="--")
plt.plot(index, y_hat, label="预测值", color="g", marker="o", ls="--")
plt.xlabel("测试集样本序号")
plt.ylabel("样本的真实值（预测值）")
plt.legend()
```
