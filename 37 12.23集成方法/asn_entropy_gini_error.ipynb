```python
import numpy as np

def entropy(p1, p2):
    # 定义一个很小的正数，防止真数为0。
    e = 1e-5
    return -p1 * np.log2(p1 + e) - p2 * np.log2(p2 + e)

def gini(p1, p2):
    return 1 - p1 ** 2 - p2 ** 2

def error(p1, p2):
    return 1 - max(p1, p2)

def compute(impurity):
    # 计算父节点不纯度。
    parent = impurity(0.5, 0.5)
    left_a = impurity(3 / 4, 1 / 4)
    right_a = impurity(1 / 4, 3 / 4)
    gain_a = parent - 0.5 * left_a - 0.5 * right_a
    
    left_b = impurity(1 / 3, 2 / 3)
    right_b = impurity(1, 0)
    gain_b = parent - 3 / 4 * left_b - 1 / 4 * right_b
    
    print(gain_a, gain_b)
    
compute(entropy)
compute(gini)
compute(error)
```
