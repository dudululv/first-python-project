```python
#创建高维索引数据
df = pd.DataFrame([
[1,2],[3,4],[5,6],[7,8]
],index=[['a','a','b','b'],['c','d','c','d']])
df
#设置level=1实现按照第2层索引分组
g1=df.groupby(level=1)
#使用for循环选代输出分组结果
for i in g1:
    print(i)
```

```python
#分组后的函数 nunique() 对每个分组的 每一列 统计唯一值数量。
df = pd.DataFrame([
[1,3],[3,4],[1,6],[7,8]
],index=[['a','a','b','b'],['c','d','c','d']])
df
df.groupby(level=1).nunique()
```

```python
#分组后的函数value_counts（） 唯一值的统计
df = pd.DataFrame([
[1,2],[3,4],[1,2],[7,8]
],index=[['a','a','b','b'],['c','d','c','d']])
df
df.groupby(level=1).value_counts()
```

```python
import pandas as pd
import numpy as np

# 创建示例数据
data = {
    'Year': [2023, 2023, 2023, 2023, 2024, 2024, 2024, 2024],
    'Quarter': ['Q1', 'Q1', 'Q2', 'Q2', 'Q1', 'Q1', 'Q2', 'Q2'],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South', 'North', 'South'],
    'Sales': [100, 150, 120, 180, 130, 160, 140, 190],
    'Profit': [20, 30, 25, 35, 28, 32, 30, 38]
}

df = pd.DataFrame(data)
print("原始数据:")
print(df)
# 设置高维索引（Year, Quarter, Region）
df_multiindex = df.set_index(['Year', 'Quarter', 'Region'])
print("\n设置高维索引后的DataFrame:")
print(df_multiindex)
```

```python
# 使用高维索引分组 - 高效方式
sales_by_year = df_multiindex.Sales.groupby('Year').sum()
print("\n按年份分组销售额（高维索引）:")
print(sales_by_year)
```

```python
# 传统分组方式（无高维索引）
traditional_grouping = df.groupby('Year')['Sales'].sum()
print("\n传统分组方式结果:")
print(traditional_grouping)

# 验证结果是否一致
print(f"\n结果是否一致: {sales_by_year.equals(traditional_grouping)}")
```
