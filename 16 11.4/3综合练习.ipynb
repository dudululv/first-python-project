```python
import pandas as pd
import numpy as np
df=pd.read_csv('..\data\player.csv')
df
#df.head(2)
# 分别为数据集设置高维索引
# weight_class:weight>170为1，否则为0
#height class:height>185为1，否则为
# 分别使用np.select和np.where完成
df['weight_class']=np.select([df['weight']>170],[1],0)
df['height_class']=np.where(df['height']>185,1,0)
df
df.set_index(['weight_class','height_class'],inplace=True)
df
```

```python
#分别按照身高分类、体重分类分组并实现对应分组平均值的聚合
df.groupby('height_class')['height'].mean()
df.groupby('weight_class')['weight'].mean()
```

```python
# 按身高分类分组后，获取身高分类为0的数据
df.groupby('height_class').get_group(0)
```

```python
#将数据按照身高分类分组，对分组结果的体重进行均值聚合
df.groupby('height_class')[['weight']].mean()
```

```python
# 将数据按照身高分类分组，实现每个球员的体重与组内均值的差
df.groupby('height_class', group_keys=True)[['weight']].apply(lambda x:x-x.mean())
```

```python
#实现按照球员出生年份分组，并对分组后的球员身高、体重实现平均值、计数聚合
df['year']=df.birthday.apply(lambda x:x[:4])
df.head(2)
df.groupby(by='year').agg({'height':'mean','weight':'mean','player_api_id':'count'}).rename(columns={'player_api_id':'人数'})
```
