# 张量

```python
import torch

# 不同维度的张量演示。
# 标量（0维）。
scaler = torch.tensor(3.0)
# 向量（1维）。
vector = torch.tensor([1, 2, 3])
# 矩阵（2维）。
matrix = torch.tensor([[1, 2], [3, 4]])
# RGB图像（3维）。
image = torch.rand(3, 224, 224)
# 视频片段（4维）。
video = torch.rand(10, 3, 224, 224)

print(scaler)
print(vector)
print("标量形状：", scaler.shape)
print("图像形状：", image.shape, image.size())
print("某个维度的长度：",  image.shape[0], image.size(0))

```

# 创建张量

```python
import numpy as np

# 从Python列表创建。
t1 = torch.tensor([1, 2, 3])
# 从ndarray数组创建张量，共享底层内存。
array = np.array([4, 5, 6])
t2 = torch.from_numpy(array)
array[0] = 100
print(array)
print(t2)
# 通过tensor创建张量，不共享底层内存（数据复制）。
array = np.array([4, 5, 6])
t3 = torch.tensor(array)
array[0] = 100
print(array)
print(t3)

# PyTorch很多操作与Numpy都很相似。
print(torch.zeros(size=(2, 3)))
print(torch.ones(size=(2, 3)))
print(torch.arange(0, 10, 2))
print(torch.linspace(0, 1, 5))

print(torch.rand(2, 3))
print(torch.randn(2, 3))
print(torch.randint(0, 10, (2, 3)))
```

# 数据类型与设备

```python
# 指定数据类型。
t1 = torch.tensor([2], dtype=torch.int16)
t2 = torch.tensor([2], dtype=torch.float32)
print(t1, t2)

# 指定设备，优先创建在GPU上，如果没有cuda环境，则创建在CPU上。
device = "cuda" if torch.cuda.is_available() else "cpu"
# 在指定的设备上创建张量。如果没有显式指定，则张量创建在CPU上。
t3 = torch.tensor([1], device=device)
# 张量所属的设备。
print(t1.device, t2.device, t3.device)


# 不同设备的张量，不能进行运算。
# print(t1 + t3)
# 张量的复制。在CPU与GPU之间复制。
t1 = t1.to("cuda")
print(t1.device)
print(t1 + t3)
```

# 张量操作

```python
t = torch.tensor([[1, 2], [3, 4]])
# 支持索引与切片。
print(t[0, 1])
print(t[:, 0:1])
# 对维度重新排列，返回原始张量的视图。
print(t.permute(1, 0))
# 张量的转置。
print(t.T, t.t())
# 内存是否连续。
t_trans = t.T
print("内存是否连续：", t.is_contiguous(), t_trans.is_contiguous())

# 改变张量的形状。
# view总是尝试返回视图，不复制数据，要求张量在内存中必须是连续的。
# reshape优先返回视图（共享底层数据），如果无法做到（数据不连续），则返回拷贝。
print(t.view(-1))
# 错误，张量在内存中并不连续。
# print(t_trans.view(-1))
print(t_trans.reshape(-1))
print(t_trans.contiguous().view(-1))

# 在指定的位置插入一个新维度，长度为1。
t = torch.tensor([1, 2, 3])
print('*'*30)
print(t.unsqueeze(0).shape)
print(t.unsqueeze(1).shape)
# 删除指定位置长度为1的维度。如果没有指定，则删除所有长度为1的维度。
t = torch.rand(1, 5, 2, 1)
print(t.shape)
print(t.squeeze(0).shape, t.squeeze(3).shape, t.squeeze().shape)
```

# 数学运算与广播

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
print(a + b)
# 张量与标量广播。
print(a + 1)
# 不同形状张量广播。
matrix = torch.ones(2, 3)
vector = torch.tensor([1, 2, 3])
print('2*3:1',matrix + vector)

mat_a = torch.rand(2, 3)
mat_b = torch.rand(3, 2)
# 矩阵乘法。
mat_mult = torch.mm(mat_a, mat_b)
mat_mult2 = mat_a @ mat_b
print('mm:',mat_mult.shape, '@:',mat_mult2.shape)

# 聚合运算。
print(torch.sum(a))
print(torch.mean(a.float()))
print(torch.max(a))
```

# 与Numpy互操作

```python
# tensor -> Numpy
t = torch.tensor([1., 2., 3.])
# 张量转换为ndarray数组，共享内存。
arr = t.numpy()
t[0] = 100
print(arr)

# Numpy -> tensor
arr = np.array([4, 5, 6])
# ndarray数组创建张量。共享内存。
t = torch.from_numpy(arr)
arr[0] = 100
print(t)

# 不共享内存。
arr = np.array([4, 5, 6])
t = torch.tensor(arr)
arr[0] = 100
print(t)

t = torch.tensor([1., 2., 3.])
arr = np.array(t)
t[0] = 100
print(arr)

t = torch.tensor(5.0)
print(type(t.numpy()), t.numpy())
# 获取张量的标量值，返回标量类型，而不返回数组类型。
v = t.item()
print(type(v), v)
```
