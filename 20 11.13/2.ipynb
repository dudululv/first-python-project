```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from seaborn import histplot,heatmap,violinplot,boxplot
df = pd.read_csv('../data/处理后的数据2.csv')
df.head(2)
```

```python
#过滤掉购买时年龄为空的数据
df=df.query('购买时年龄.notna()')
df.购买日期=pd.to_datetime(df.购买日期)
#添加几个列 年,月,季度,周,日
df=df.assign(
年=df.购买日期.dt.year,
月=df.购买日期.dt.month,
日=df.购买日期.dt.day,
季度=df.购买日期.dt.quarter,
周=df.购买日期.dt.weekofyear
)
```
