```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from seaborn import histplot,heatmap,violinplot,boxplot
df= pd.read_csv('../data/处理后的数据.csv')
df
```

```python
a=df.query('性别<2')
boxplot(a,x='性别',y='购买时年龄')
```

```python
#  实现不同性别的婴儿对于一级分类的购买数量堆积柱状图
a.groupby(['一级分类','性别']).购买数量.sum().unstack().plot\
    .bar(stacked=True)
```

```python
b=a.groupby(['二级分类','性别']).购买数量.sum().unstack()
#b.sum(axis=1)
b.assign(合计=b.sum(axis=1)).sort_values('合计',ascending=False.head(20).d)
```
