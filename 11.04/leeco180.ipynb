```python
import numpy as np
import pandas as pd

data = [
    [1, '2017-01-01', 10],
    [2, '2017-01-02', 109],
    [3, '2017-01-03', 150],
    [4, '2017-01-04', 90],
    [5, '2017-01-05', 200],
    [6, '2017-01-06', 180],
    [7, '2017-01-07', 170]
]

stadium = pd.DataFrame(data, columns=['id', 'visit_date', 'people']).astype({'id':'int64','visit_date':'str','people':'int64'})

# 1. 筛选出人数>=100的记录
a = stadium.query('people >= 100')

# 2. 计算连续序列
a['rk'] = a['id'] - a['id'].rank(method='min')

# 3. 统计连续出现 rk 值的数量
b = a.groupby('rk')['id'].count()

# 4. 找出连续出现人数>=100的序列长度>=3 的 rk 值
result = a.loc[a['rk'].isin(b[b >= 3].index), a.columns[:-1]].sort_values('visit_date')
```
