```python
import pandas as pd
import numpy as np
#生成数据
df=pd.DataFrame([
[2,5,9],
[2,10,9],
[1,2,6],
[1,5,6],
[1,2,3],
[2,5,8]
],columns=['a','b','c']) 
df
df.groupby(['a','b'])
```

```python
gab=df.groupby(['a','b'])
for i in gab:
    print(i)
```

```python
#创建高维索引数据
df = pd.DataFrame([
[1,2],[3,4],[5,6],[7,8]
],index=[['a','a','b','b'],['c','d','c','d']])
df
#设置level=1实现按照第2层索引分组
g1=df.groupby(level=1)
#使用for循环选代输出分组结果
for i in g1:
    print(i)
```

```python
#分组后的属性groupsdf
df
df.groupby(level=1).groups
```

```python
#分组后的函数df
df
df.groupby(level=1).get_group('c')
```

```python
#分组后的函数first（)每组的第一条数据last（O每组的最后一条数据
df
df.groupby(level=1).first()
df.groupby(level=1).last()
```

```python

#分组后的函数 size每组数据的个数
df
df.groupby(level=1).size()
```

```python
#分组后的函数 nunique() 对每个分组的 每一列 统计唯一值数量。
df = pd.DataFrame([
[1,3],[3,4],[1,6],[7,8]
],index=[['a','a','b','b'],['c','d','c','d']])
df
df.groupby(level=1) .nunique()
```

```python
#分组后的函数value_counts（） 唯一值的统计
df = pd.DataFrame([
[1,2],[3,4],[1,2],[7,8]
],index=[['a','a','b','b'],['c','d','c','d']])
df
df.groupby(level=1).value_counts()
```

```python
#高维索引I下的分组操作# 创建DataFrame
df=pd.DataFrame({
'a':[1,1,1,2,2,2],
'b':[2,5,2,10,5,5],'c' :[3,6,6,9,9,8]
})
#df
#设置高维索引
df.set_index(['a','b'],inplace=True)
df
#对比具有高维索引|的情况与没有高维索引情况下的分组操作
#有高维索引
df.c.groupby(level=0).groups#没有高维索引
df.reset_index('b',inplace=True)
df
df.c.groupby(level=0).groups
```
