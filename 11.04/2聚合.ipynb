```python
import pandas as pd
import numpy as np
np.random.seed(0)
a=pd.DataFrame(np.random.randint(10,100,(20,4)),
               columns=['S0','S1','S2','S3'],
               )
a
a['I1']=[0,1]*10
a
a.set_index('I1',inplace=True)
a
```

```python
a.groupby(level=0).mean()
```

```python
a.groupby(level=0).sum()
a.groupby(level=0).sum()['S1']
a.groupby(level=0)['S1'].sum()
```

```python
a.reset_index(inplace=True)
a
```

```python
a.groupby(by='I1').mean()
```

```python
a.groupby(by='I1')['S1'].sum()
```

```python
#滚动窗口聚合rolling()
a.groupby(by='I1').rolling(5).mean()
# 最小窗口为1
#min_periods=1()
a.groupby(by='I1').rolling(5,min_periods=1).mean().round(2)
```

```python
a.reset_index(inplace=True)
```

```python
b=a.groupby(by='I1').transform('mean').rename(columns={'S0':'S0_mean','S1':'S1_mean','S2':'S2_mean','S3':'S3_mean'})
b

a=a.merge(b,how='inner',)
a
```

```python
#apply
import pandas as pd
import numpy as np
#生成数据
df=pd.DataFrame([
[2,5,9],
[2,10,9],
[1,2,6],
[1,5,6],
[1,2,3],
[2,5,8]
],columns=['a','b','c']) 
df
df.groupby(['a']).mean()
df.groupby('a').mean()
df.groupby('a',group_keys=False).apply(lambda x:x-x.mean())
df.groupby('a',group_keys=False)['b'].apply(lambda x:x-x.mean())
```

```python
df.agg('mean')
df.agg(np.mean)
df.agg(['mean','max'])
df.agg([np.mean,np.max])
df.agg({'b':np.mean,'c':'max'})
```
