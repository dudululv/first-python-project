```python
#df创建
import pandas as pd
import numpy as np

#1 二维列表创建
d=[
    [1,'john',18,'male','beijing'],
    [2,'susan',19,'female','shanghai'],
    [3,'gump',20,'male','newyork']
]
pd.DataFrame(data=d,index=[1,2,3],columns=['id','name','age','gender','adress'])
```

```python
#2 字典创建
dic1={
    1:[1,'john',18,'male','beijing'],
    2:[2,'susan',19,'female','shanghai'],
    3:[3,'gump',20,'male','newyork']
}
pd.DataFrame(dic1,index=['id','name','age','gender','adress']).T
```

```python
# 3 Series创建
name=pd.Series(['john','susan','gump'])
name
age=pd.Series([18,19,20])
age
pd.DataFrame([name,age]).T
```
