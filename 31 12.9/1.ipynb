```python
from typing import Any


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "SimHei"
plt.rcParams["axes.unicode_minus"] = False

def fun(x):
    """定义目标函数。
    """
    return x ** 2 - 2 * x + 1

def grad_f(x):
    """目标函数的梯度函数，返回目标函数在点x处的梯度值。
    """
    return 2 * x - 2

def grad_descent(f, g, x_init, eta=0.1, tol=1e-4, max_iter=100, verbose=True):
    """执行梯度下降，求解目标函数的最小值。
    
    Parameters
    ----------
    f : function
        目标函数。
    g : function
        目标函数的梯度函数。
    x_init : float
        初始点的位置。
    eta : float
        学习率。
    tol : float
        容忍度。在梯度法执行过程中，如果y的下降幅度，比tol还要小，则停止迭代。
    max_iter : int
        最大迭代次数。
    verbose : bool
        是否显示更加详尽的信息。
    """
    # 定义列表，用来存储梯度法执行过程中，x与y的更新轨迹。
    x_list = []
    y_list = []
    x = x_init
    for i in range(max_iter):
        x_list.append(x)
        y = f(x)
        y_list.append(y)
        # 如果当前的y，比上一次迭代的y值，下降幅度小于tol，则停止迭代。
        if len(y_list) >= 2 and y_list[-2] - y < tol:
            break
        # 根据负梯度的方向，更新自变量，使得函数值越来越小。
        x = x - eta * g(x)
    if verbose:
        for index, (a, b) in enumerate(zip(x_list, y_list), start=1):
            print(f"第{index}次迭代 x={a} y={b}")
    return x_list, y_list


x_list, y_list = grad_descent(fun, grad_f, x_init=10)
```

```python
def plot_result(f, x, x_li, y_li):
    """绘制梯度法更新，自变量与函数的更新轨迹。
    
    Parameters
    ----------
    f : function
        目标函数。
    x : array
        目标函数f的定义域。
    x_li : list
        自变量的更新轨迹。
    y_li : list
        函数值的更新轨迹。
    """
    y = f(x)
    plt.plot(x, y, c="y")
    plt.plot(x_li, y_li, marker="o", ls="--", c="r")
    plt.plot(x_li[0], y_li[0], marker="*", ms=15, c="b", label="起始点")
    plt.plot(x_li[-1], y_li[-1], marker="*", ms=15, c="g", label="终止点")
    plt.legend()
    
domain = np.linspace(-9, 11, 200)
plot_result(fun, domain, x_list, y_list)


```

```python
def fun(x1, x2):
    """目标函数。
    """
    return 0.2 * (x1 + x2) ** 2 - 0.3 * x1 * x2 + 0.4

def grad_f_x1(x1, x2):
    """目标函数在(x1, x2)点，对x1求梯度（偏导）。
    """
    return 0.4 * (x1 + x2) - 0.3 * x2

def grad_f_x2(x1, x2):
    """目标函数在(x1, x2)点，对x2求梯度（偏导）。
    """
    return 0.4 * (x1 + x2) - 0.3 * x1

def grad_descent(f, g, x_init, eta=0.1, tol=1e-4, max_iter=100, verbose=True):
    x_list = []
    y_list = []
    x1, x2 = x_init
    g1, g2 = g
    for i in range(max_iter):
        x_list.append((x1, x2))
        y = f(x1, x2)
        y_list.append(y)
        if len(y_list) >= 2 and y_list[-2] - y < tol:
            break
        v1 = g1(x1, x2)
        v2 = g2(x1, x2)
        x1 -= eta * v1
        x2 -= eta * v2
    if verbose:
        for index, (a, b) in enumerate(zip(x_list, y_list), start=1):
            print(f"第{index}次迭代， x={a} y={b}")
    return x_list, y_list

x_list, y_list = grad_descent(fun, g=(grad_f_x1, grad_f_x2), x_init=(4.8, 4.5), eta=0.4)
```

```python
from mpl_toolkits.mplot3d import Axes3D
%matplotlib qt

def plot_result(f, x, x_li, y_li):
    x1, x2 = x
    X1, X2 = np.meshgrid(x1, x2)
    X = np.array([X1.ravel(), X2.ravel()]).T
    y = f(X[:, 0], X[:, 1])
    Y = y.reshape(X1.shape)
    fig = plt.figure()
    ax = Axes3D(fig)
    fig.add_axes(ax)
    surf = ax.plot_surface(X1, X2, Y, rstride=5, cstride=5, cmap=plt.cm.rainbow, alpha=0.7)
    # x_li 是一个二维列表，不方便取列，为了方便取列，我们将其转换为ndarray数组。
    array = np.asarray(x_li)
    x1_trace = array[:, 0]
    x2_trace = array[:, 1]
    ax.plot(x1_trace, x2_trace, y_li, c="b", ls="--", marker="o")
    ax.plot(x1_trace[0], x2_trace[0], y_li[0], marker="*", ms=15, label="起始点", c="b")
    ax.plot(x1_trace[-1], x2_trace[-1], y_li[-1], marker="*", ms=15, label="终止点", c="g")
    fig.colorbar(surf)
    ax.legend()
    
    fig2 = plt.figure()
    ax2 = fig2.gca()
    # 绘制填充的等高线。
    # levels：设置登高线的疏密程度，该值越大，则等高线越密。
    m = ax2.contourf(X1, X2, Y, levels=30)
    ax2.scatter(x1_trace, x2_trace, c="r")
    fig2.colorbar(m)
    
x1_domain = np.linspace(-5, 5, 100)
x2_domain = np.linspace(-5, 5, 100)
plot_result(fun, (x1_domain, x2_domain), x_list, y_list)
```
